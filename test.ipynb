```python
import random

class env:
    def __init__(self, length, seed, max_timesteps):
        self.length = length
        self.seed = seed
        self.max_timesteps = max_timesteps
        self.actions = {0, 1}
        random.seed(self.seed)
        self._create()

        
    def _create(self):
        self.done, self.truncated = False, False
        self.timestep = 1
        self.rewards = [0] * self.length
        self.rewards[0], self.rewards[-1] = -1, 1
        self.agent_position = random.randint(1, self.length-1)

    def _evaluate(self):
        if self.agent_position == 0 or self.agent_position == self.length-1:
            self.done = True
        if self.timestep >= self.max_timesteps:
            self.done = True
            self.truncated = True
    
    def reset(self):
        self._create()
        return self.agent_position, {"length":self.length, "rewards":self.rewards, "agent position":self.agent_position, "done": self.done, "truncated":self.timestep >= self.max_timesteps}

    def step(self, action):
        if action not in self.actions:
            raise ValueError("Undefiend Action.")

        if action == 0 and self.agent_position > 0:
            self.agent_position -= 1
        elif action == 1 and self.agent_position < self.length - 1:
            self.agent_position += 1

        self._evaluate()
        return self.agent_position, self.rewards[self.agent_position], self.done, self.truncated , {"length":self.length, "rewards":self.rewards, "agent position":self.agent_position, "done": self.done, "truncated":self.timestep >= self.max_timesteps}
    

    
```

```python
env = env(length=10, seed=42, max_timesteps=20)
```
